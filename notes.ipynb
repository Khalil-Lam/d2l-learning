### 1.1.5运算的内存开销

我们对每个操作新开内存来存储运算结果 如Y=Y+X 会开新内存 并把新的Y指向新内存

这点我们可以用python自带的id函数验证，若id同 则地址同

```python
Before=id(y)
y=y+4
id(y)==Before
```

pytorch这么写

```python
import torch
before = id(y)
y = y + x
id(y) == before
```

* 为什么要关心这个？

    因为大张量很占内存。

    如果每次都新建一个张量，旧的还没释放，新的一直开，内存压力会变大。

方法 1：提前准备一个 Z，把结果写进去

(利用索引的替换操作把指定结果放到指定内存 先通过zero-like创建和y形状相同的tensor，记作z 再把X+Y的结果通过[:]写进z对应的内存中)

```python
Z = torch.zeros_like(y)

before = id(Z)
Z[:] = x+ y

id(Z) == before
```

意思是：

Y = Y + X 会创建一个新的张量，然后让变量 Y 指向这个新张量。

也就是说，原来的 Y 没有被直接修改，而是换了一个新的对象。

方法 2：用 out=，更直接

PyTorch 写法：

```python
Z = torch.zeros_like(y)

before = id(Z)
torch.add(x, y, out=Z)

id(Z) == before
```

意思是：

直接把 X + Y 的结果写进 Z，尽量避免额外创建结果张量

方法 3：直接原地改 X

如果后面不再需要旧的 X，可以写：

```python
before = id(x)
x += y
id(x) == before
```

这表示：

X += Y 会直接在原来的 X 上修改。旧的 X 没了

* PyTorch 里很多带下划线的操作，比如 add_()，表示原地操作；官方文档也把 Tensor.add_() 定义为 add() 的 in-place 版本。

| 方法   | PyTorch 写法               |  是否换内存 | 是否推荐 | 含义                  |
| ---- | ------------------------ | -----: | ---- | ------------------- |
| 方法 1 | `Z[:] = X + Y`           | `Z` 不换 | 可以   | 先算出 `X + Y`，再写进 `Z` |
| 方法 2 | `torch.add(X, Y, out=Z)` | `Z` 不换 | 更好   | 直接把结果写进 `Z`，更省临时内存  |
| 方法 3 | `X += Y` 或 `X.add_(Y)`   | `X` 不换 | 谨慎   | 直接修改 `X` 本身         |



| 写法                       | 会不会新建变量对象 |     会不会覆盖原数据 | 适合场景           |
| ------------------------ | --------: | -----------: | -------------- |
| `Y = Y + X`              |         会 | 不覆盖旧数据，只是换引用 | 普通写法，最直观       |
| `Z[:] = X + Y`           |   不会换 `Z` |       覆盖 `Z` | 想把结果放进已有 `Z`   |
| `torch.add(X, Y, out=Z)` |   不会换 `Z` |       覆盖 `Z` | 更省内存地把结果写进 `Z` |
| `X += Y`                 |   不会换 `X` |       覆盖 `X` | 不再需要旧 `X` 时    |

## 1.2自动求梯度

```python
import torch

x = torch.arange(4.0).reshape(4, 1)
x.requires_grad_(True)

y = 2 * torch.matmul(x.T, x)

y.backward()

print(x.grad)
```

* “x = torch.arange(4.0).reshape(4, 1)”

  这里4.0
  因为求梯度要求浮点数张量。整数张量不能求梯度，PyTorch 文档也说明自动微分支持浮点和复数 Tensor。

x.requires_grad_(True)  意思是：

* 从现在开始，PyTorch 要记录和 x 有关的计算，后面准备对 x 求梯度。

y = 2 * torch.matmul(x.T, x)

* 不仅仅是算出来数 也记录了算出数的方式，方便后续的反向传播
* 一言概之：生成 y 和计算图

y.backward()

* 这句的意思是：

    从 y 开始，反向计算 y 对 x 的梯度。

* PyTorch 文档说明，backward() 会计算当前张量相对于计算图中叶子节点的梯度，并累积到这些张量的 .grad 属性里。

* y.backward计算 算完放x.grad里

* 注意：反向传播只能算一次，因为计算过程中，算完直接把原来的函数的信息“烧”掉了，以节省显存，要再次看就要重启kernel


### 1.2.1对python控制流求梯度

* 就算函数里面有 while、if 这种 Python 控制流，PyTorch 也能自动求梯度。

  因为 PyTorch 的 autograd 会在你实际执行代码时记录计算图，然后用链式法则反向求导；这个计算图是动态生成的，所以 Python 的条件判断、循环也能参与计算过程。



1. 先定义一个带控制流的函数


```python
import torch

def f(a):
    b = a * 2
    
    while b.norm() < 1000:  #这里norm一般指L2范数
        b = b * 2
    
    if b.sum() > 0:
        c = b
    else:
        c = 100 * b
    
    return c
```

这个函数里面有两个控制流：

* while 循环

* if 判断

```python
a = torch.randn(1, requires_grad=True)

c = f(a)

c.backward()

a.grad
```

2. 为什么能求导？

   每次真正运行时，PyTorch 都会记录实际发生过的计算。

   PyTorch 会记录这条实际路径。

   然后调用：

   c.backward()

   它就沿着这条路径反向求导。

3.验证

```python
import torch

def f(a):
    b = a * 2
    
    while b.norm() < 1000:
        b = b * 2
    
    if b.sum() > 0:
        c = b
    else:
        c = 100 * b
    
    return c


a = torch.randn(1, requires_grad=True)
c = f(a)

c.backward()

print("a =", a)
print("c =", c)
print("a.grad =", a.grad)
print("c / a =", c / a)
print(torch.allclose(a.grad, c / a))
```

* 解释

    a = torch.randn(1, requires_grad=True)

    生成一个随机数，并且告诉 PyTorch：

    我要对 a 求梯度。

    这里的 requires_grad=True 很关键。没有它，PyTorch 不会记录 a 后面的计算。官方文档也说，autograd 会追踪 requires_grad=True 的 Tensor 参与的操作。

7. 注意一个小坑

    如果你多次运行：
    c.backward()

    梯度会累加到：
    a.grad

    （例如）原来的 a.grad = 4
    这次又算出来 4
    累加后 = 8

    PyTorch 的 backward() 会把梯度累积到叶子张量的 .grad 里，所以训练时一般要先清零梯度。

    比如：a.grad.zero_()

    或者训练模型时常见：

    optimizer.zero_grad()

    否则梯度会越加越多，像你没清空购物车一样，越看越离谱。
* 正确做法：每次 backward 前先清零

```python
import torch

a = torch.tensor(2.0, requires_grad=True)

c = a ** 2
c.backward()
print(a.grad)   # tensor(4.)

 # 无清空梯度

c = a ** 2
c.backward()
print(a.grad)   # tensor(4.)
```

```python
import torch

a = torch.tensor(2.0, requires_grad=True)

c = a ** 2
c.backward()
print(a.grad)   # tensor(4.)

a.grad.zero_()  # 清空梯度

c = a ** 2
c.backward()
print(a.grad)   # tensor(4.)
```

8. 为什么要这样设计？

因为有些时候我们确实想累加梯度。

比如一个大 batch 太大，显存放不下，可以拆成几个小 batch：

小 batch 1 算一次梯度

小 batch 2 算一次梯度

小 batch 3 算一次梯度

最后一起更新参数

这叫 梯度累积 gradient accumulation。

所以 PyTorch 默认让梯度累加，不擅自清空。它不替你做主，像个冷漠但讲规矩的会计。

## 2.1线性回归

### 2.1.1线性回归基本要素

* 0. 先理解：什么是前向？什么是后向


      0.1前向传播就是：

         把输入数据丢进模型，算出预测值，再算出损失。

         在线性回归里：

            y_hat = X @ w + b

            loss = squared_loss(y_hat, y)

          所以前向传播做三件事：

            输入 X
            ↓
            模型算预测值 ŷ
            ↓
            损失函数算 loss



      0.2 后向传播就是：

         从损失 loss 出发，反过来计算每个参数对 loss 的影响，也就是梯度。

         在 PyTorch 里就是：

            loss.backward()


      0.3 一句话：

             前向传播：从输入到损失。

             后向传播：从损失反推梯度。
